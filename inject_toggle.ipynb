```python
import os
import re

button_html = '\n            <button class="theme-toggle-btn" id="themeToggleBtn" aria-label="Toggle Theme">\n                <i id="themeToggleIcon" class="bx bx-moon"></i>\n            </button>'

def process_file(filepath):
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            content = f.read()
    except UnicodeDecodeError:
        try:
            with open(filepath, 'r', encoding='latin-1') as f:
                content = f.read()
        except Exception as e:
            print(f"Error reading {filepath}: {e}")
            return False
    
    # Match <div class="nav-actions ...">
    pattern = re.compile(r'(<div\s+[^>]*class=["\']nav-actions["\'][^>]*>)')
    
    match = pattern.search(content)
    if not match:
        return False
        
    # Check if already has the button
    if 'themeToggleBtn' in content:
        print(f"Already contains toggle button: {filepath}")
        return False
        
    # Insert button after the matching tag
    pos = match.end()
    new_content = content[:pos] + button_html + content[pos:]
    
    try:
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write(new_content)
        print(f"Successfully added button to: {os.path.basename(filepath)}")
        return True
    except Exception as e:
        print(f"Error writing to {filepath}: {e}")
        return False

# Run injection
root_dir = r"d:\InternShip Project\VGB-Banking-System-Java\src\main\webapp"
for root, dirs, files in os.walk(root_dir):
    for file in files:
        if file.endswith('.jsp'):
            filepath = os.path.join(root, file)
            process_file(filepath)
```
